                                                        Exploratory Data Analysis on a Dataset 

```python
#Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
df=pd.read_csv("Cardiotocographic.csv")
df.head()
```

```python
df.info()
```

```python
#Handle Missing Values
df.isnull().sum()
df.fillna(df.mean(), inplace=True)
```

```python
#Check Data Types
df.dtypes
```

```python
#Detect Outliers
plt.figure(figsize=(10,5))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
df.mean()
```

```python
df.median()
```

```python
df.std()
```

```python
#Histograms
df.hist(figsize=(15,10))
plt.tight_layout()
plt.show()
```

```python
#Boxplots for Each Feature
for col in df.columns:
    plt.figure()
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
#Correlation heatmap
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Scatter Plot to find relationship between features
sns.scatterplot(x='LB',y='AC',data=df)
plt.title("LB vs AC")
plt.show()
```

```python
#Pairplot for deeper insights
sns.pairplot(df)
plt.show()
```

```python
#Distribution Visualization using Violin Plot
plt.figure(figsize=(10,5))
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

## Key Insights
- Most variables show normal or slightly skewed distribution.
- Some features contain outliers which may affect analysis.
- Correlation heatmap shows relationships between fetal health indicators.
- Features like LB, AC, and UC show meaningful patterns.


- The dataset provides useful insights into fetal health.
- Variability features (ASTV, MSTV, ALTV, MLTV) are important indicators.
- Strong correlations between some features suggest possible predictive modeling.
- Further analysis like machine learning can be applied for classification.

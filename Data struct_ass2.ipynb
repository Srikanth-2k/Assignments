```python
my_list = [5, 2, 9, 1, 7]
print("Original list:", my_list)
```

```python
# append
my_list.append(10)
print("After append:", my_list)
```

```python
# extend
my_list.extend([11, 12])
print("After extend:", my_list)
```

```python
# remove
my_list.remove(9)
print("After remove:", my_list)
```

```python
# insert
my_list.insert(2, 100)
print("After insert:", my_list)
```

```python
my_list.pop()
print("After pop:", my_list)
```

```python
# index
print("Index of 100:", my_list.index(100))
```

```python
# count
print("Count of 1:", my_list.count(1))
```

```python
# sort
my_list.sort()
print("After sort:", my_list)
```

```python
# reverse
my_list.reverse()
print("After reverse:", my_list)
```

```python
# clear
my_list.clear()
print("After clear:", my_list)
```

6) Given a 'marks' variable, print the grade: - 90 and above: 'A' - 75–89: 'B' - 50–74: 'C' - Below 50: 'Fail'

```python
marks=82
if marks>=90:
    print("A")
elif marks>=75:
    print("B")
elif marks>=50:
    print("C")
else:
    print("Fail")
```

7) Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
num=248
if num>0:
    print("Positive number")
    if num %2==0:
        print("It is also even")
    else:
        print("It is odd")
elif num==0:
    print("Zero")
else:
    print("Negative number")
```

8) Create: 
- A scalar using np.array(5) 
- A 1D array with values 1 to 5 
- A 2D array (2x3) with values from 10 to 60 in steps of 10 

```python
import numpy as np
import pandas as pd

# scalar
scalar=np.array(5)
print("Scalar:", scalar)

# 1D array
arr1=np.array([21,78,34,4,95])
print("1D array:", arr1)

# 2D array (2x3)
arr2 = np.array([[10,20,30], [40,50,60]])
print("2D array:\n", arr2)
```

                                                        Estimation And Confidence Intervals 

```python
import numpy as np
import scipy.stats as stats
```

```python
data = np.array([
    1.13,1.55,1.43,0.92,1.25,
    1.36,1.32,0.85,1.07,1.48,
    1.20,1.33,1.18,1.22,1.29])
```

```python
n=len(data)                    
mean=np.mean(data)           
std_sample=np.std(data,ddof=1)
print("Sample Size:",n)
print("Sample Mean:",round(mean,4))
print("Sample Standard Deviation:",round(std_sample,4))
```

PART-A

```python
confidence_level=0.99
```

```python
df=n-1
```

```python
t_critical=stats.t.ppf((1+confidence_level)/2,df)
```

```python
margin_error_t=t_critical*(std_sample/np.sqrt(n))
```

```python
ci_lower_t=mean-margin_error_t
ci_upper_t=mean+margin_error_t
```

```python
print("\n--- Part (a): Using t-distribution ---")
print("t-critical value:",round(t_critical,4))
print("Margin of Error:",round(margin_error_t,4))
print("99% Confidence Interval:({:.4f},{:.4f})".format(ci_lower_t,ci_upper_t))
```

PART-B

```python
sigma=0.2
```

```python
z_critical=stats.norm.ppf((1+confidence_level)/2)
```

```python
margin_error_z=z_critical*(sigma/np.sqrt(n))
```

```python
ci_lower_z=mean - margin_error_z
ci_upper_z=mean + margin_error_z
```

```python
print("\n--- Part (b):Using z-distribution ---")
print("z-critical value:",round(z_critical,4))
print("Margin of Error:",round(margin_error_z,4))
print("99% Confidence Interval:({:.4f},{:.4f})".format(ci_lower_z,ci_upper_z))
```

                                                        Hypothesis Testing 

```python
import numpy as np
import scipy.stats as stats
```

```python
sample_mean=3050
n=25   
```

```python
mu_X=600           
sigma_X=25
```

```python
mu_theoretical=1000+5 * mu_X
```

```python
sigma_W=5*sigma_X
```

```python
print("Theoretical Mean Cost (μ):",mu_theoretical)
print("Standard Deviation of Cost (σ):",sigma_W)

```

```python
print("\n--- Hypothesis Statements ---")
print("Null Hypothesis (H0):Actual mean cost=Theoretical mean cost")
print("Alternative Hypothesis (H1):Actual mean cost >Theoretical mean cost")
```

```python
z=(sample_mean - mu_theoretical)/(sigma_W / np.sqrt(n))
print("\nTest Statistic (Z):",round(z,4))
```

```python
alpha=0.05
```

```python
z_critical=stats.norm.ppf(1 - alpha)
print("Critical Value (Zα):",round(z_critical,4))
```

```python
print("\n--- Decision ---")
if z > z_critical:
    print("Reject the Null Hypothesis (H0)")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
```

```python
print("\n--- Conclusion ---")
if z > z_critical:
    print("There is strong evidence that the weekly operating cost is higher than the model suggests.")
else:
    print("There is not enough evidence to say that the weekly operating cost is higher than the model.")
```

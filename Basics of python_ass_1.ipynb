Exercise 1: Prime Numbers 

```python
#checks whether the number is prime or not 

def prime_number(num):
    if num<=1:
        return False
    for i in range(2,int(num)):
        if num%i ==0:
            return False
    return True

number=int(input("Enter the number:"))
if prime_number(number):
    print(f"{number} is a prime number")
else:
    print(f"{number} is not a prime number") 
```

Exercise 2: Product of Random Numbers

```python
import random

num1=random.randint(1,30)
num2=random.randint(1,20)

product=num1*num2

user_input=int(input(f" what is the product of {num1} and {num2}:"))

if user_input==product:
    print(f" Correct,the product of {num1} and {num2} is,{product}.")
else:
    print(f" sorry the product of {num1} and {num2} is, {product}.")    
```

Exercise 3: Squares of Even/Odd Numbers 

```python
#prints the square of all odd numbers between 100 to 200

for num in range(100,200):
    if num%2!=0: #checks the number is odd
        square= num**2 #calculates the square of the number
        print(f"{square} is square of {num}")

```
